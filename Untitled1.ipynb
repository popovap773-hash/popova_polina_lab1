```python
import numpy as np
import matplotlib.pyplot as plt


def print_theory():

    print("Выражения для безразмерной энергии e(ϕ) = 1/2 * sin²(ϕ) - h * cos(θ - ϕ):")
    print("1) При θ = 0°:   e(ϕ) = 1/2 * sin²(ϕ) - h * cos(ϕ)")
    print("2) При θ = 45°:  e(ϕ) = 1/2 * sin²(ϕ) - h * cos(45° - ϕ)")
    print("3) При θ = 90°:  e(ϕ) = 1/2 * sin²(ϕ) - h * sin(ϕ)")

    print("Формула критического поля (астроиды):")
    print("h_crit(θ) = (sin^(2/3)(θ) + cos^(2/3)(θ))^(-3/2)")
    print("Для θ = 45°: h_crit(45°) = ( (1/√2)^(2/3) + (1/√2)^(2/3) )^(-3/2) = 0.5")
    print("\n")


def calculate_hysteresis(theta_deg, h_max=2.0, n_points=1000):

    theta = np.radians(theta_deg)


    h_down = np.linspace(h_max, -h_max, n_points)
    h_up = np.linspace(-h_max, h_max, n_points)
    h_sweep = np.concatenate([h_down, h_up])

    m_sweep = np.zeros_like(h_sweep)


    phi = theta

    for i, h in enumerate(h_sweep):

        for _ in range(100):

            f = 0.5 * np.sin(2 * phi) - h * np.sin(theta - phi)

            df = np.cos(2 * phi) + h * np.cos(theta - phi)

            # Проверка на знаменатель
            if abs(df) < 1e-8:
                phi += 0.05
                continue

            dphi = f / df
            phi -= dphi


            if abs(dphi) < 1e-6:
                break


        df = np.cos(2 * phi) + h * np.cos(theta - phi)

        if df <= 0:

            phi_scan = np.linspace(-np.pi, np.pi, 10000)
            e = 0.5 * np.sin(phi_scan)**2 - h * np.cos(theta - phi_scan)
            phi = phi_scan[np.argmin(e)]


        m_sweep[i] = np.cos(theta - phi)

    return h_sweep, m_sweep

def plot_hysteresis_loops():
    fig, ax = plt.subplots(figsize=(8, 6))
    angles = [0, 30, 60, 90]
    colors = ['red', 'blue', 'green', 'purple']

    for theta, color in zip(angles, colors):
        h, m = calculate_hysteresis(theta)
        ax.plot(h, m, label=f'θ = {theta}°', color=color, linewidth=2)

    ax.set_title('Петли гистерезиса m(h) для различных углов θ', fontsize=14)
    ax.set_xlabel('Безразмерное поле h', fontsize=12)
    ax.set_ylabel('Намагниченность m', fontsize=12)
    ax.set_xlim([-2, 2])
    ax.set_ylim([-1.1, 1.1])
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.legend()
    plt.tight_layout()
    plt.show()

def plot_astroid():
    fig, ax = plt.subplots(figsize=(7, 7))

    phi = np.linspace(0, 2 * np.pi, 1000)
    hx = -np.sin(phi)**3
    hz = np.cos(phi)**3

    ax.plot(hx, hz, 'b-', linewidth=2, label='Астроида Стонера-Вольфарта')

    ax.fill(hx, hz, color='cyan', alpha=0.2, label='Область бистабильности')

    ax.set_title('Астроида Стонера-Вольфарта', fontsize=14)
    ax.set_xlabel('$h_x$ (перпендикулярно ОЛН)', fontsize=12)
    ax.set_ylabel('$h_z$ (вдоль ОЛН)', fontsize=12)
    ax.set_aspect('equal')
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.legend()
    plt.tight_layout()
    plt.show()

def plot_hc_vs_theta():
    fig, ax = plt.subplots(figsize=(8, 5))

    theta_deg = np.linspace(0, 90, 200)
    theta_rad = np.radians(theta_deg)

    # (13)
    hc = (np.sin(theta_rad)**(2/3) + np.cos(theta_rad)**(2/3))**(-1.5)

    ax.plot(theta_deg, hc, 'r-', linewidth=2, label='$h_c(θ)$')

    # θ = 45°
    ax.plot(45, 0.5, 'ko', markersize=8, label='$h_c(45°) = 0.5$')
    ax.annotate('0.5', xy=(45, 0.5), xytext=(55, 0.6),
                arrowprops=dict(facecolor='black', shrink=0.05, width=1.5, headwidth=8))

    ax.set_title('Зависимость коэрцитивной силы от угла θ', fontsize=14)
    ax.set_xlabel('Угол θ (градусы)', fontsize=12)
    ax.set_ylabel('Коэрцитивная сила $h_c$', fontsize=12)
    ax.set_xlim([0, 90])
    ax.set_ylim([0, 1.1])
    ax.grid(True, linestyle='--', alpha=0.7)
    ax.legend()
    plt.tight_layout()
    plt.show()


def print_analysis():

    print("1. Причина исчезновения гистерезиса при θ = 90°:")
    print("При θ = 90° поле направлено перпендикулярно оси легкого намагничивания.")
    print("Уравнение равновесия принимает вид: cos(ϕ) = h.")
    print("Процесс перемагничивания происходит путем обратимого вращения")
    print("вектора намагниченности. Энергетические барьеры между состояниями")
    print("отсутствуют, поэтому гистерезис не наблюдается (hc = 0).")

    print("2. Сравнение коэрцитивной силы при θ = 45°:")
    print("Согласно теоретической формуле: h_c(45°) = 0.5.")
    print("На графике зависимости hc(θ) это значение подтверждается.")
    print("Это минимальное значение коэрцитивной силы для углов 0 < θ < 90°.")



if __name__ == "__main__":
    # Вывод теоретических выкладок
    print_theory()

    # Построение графиков
    print("Построение петель гистерезиса...")
    plot_hysteresis_loops()

    print("Построение астроиды Стонера-Вольфарта...")
    plot_astroid()

    print("Построение графика зависимости hc от θ...")
    plot_hc_vs_theta()

    # Вывод анализа
    print_analysis()
```
